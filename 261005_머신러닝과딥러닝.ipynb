# 머신 러닝과 딥러닝
## 인공지능, 머신 러닝과 딥러닝

*   인공지능(Artificial Intelligence, AI)은 인간의 지능을 모방하여 사람이 하는 일을 컴퓨터(기계)가 할 수 있도록 하는 기술임. 인공지능을 구현하는 방법으로는 머신 러인과 딥러닝이 있음.
* 인공지능과 머신러닝, 딥러닝의 관계: 인공지능 > 머신 러닝 > 딥러닝
* **머신 러닝과 딥러닝 모두 학습 모델을 제공해 데이터를 분류할 수 있는 기술이나, 접근 방식에 차이가 있음.** 머신 러닝은 주어진 데이터를 인간이 먼저 처리(전처리)함. **머신 러닝은 범용적인 목적을 위해 제작된 것으로 데이터의 특징을 스스로 추출하지 못하며, 이 과정을 인간이 처리해 주어야 함.** (e.g. 이미지 데이터라면 사람이 학습(train) 데이터를 컴퓨터가 인식할 수 있도록 준비해두어야 함.) 반면 **딥러닝은 인간이 하던 작업을 생략하고, 대량의 데이터를 신경망에 적용하면 컴퓨터가 스스로 분석한 후 답을 찾음.**

## 머신 러닝이란
### 머신 러닝 학습 과정

*   머신 러닝은 크게 학습 단계(learning)와 예측 단계(prediction)로 구분할 수 있음. 훈련 데이터를 머신 러닝 알고리즘에 적용하여 학습시키고, 이 학습 결과로 모형이 생성됨. 예측 단계에서는 학습 단계에서 생성된 모형에 새로운 데이터를 적용해 결과를 예측함.
*   **머신 러닝의 주요 구성 요소는 데이터와 모델임.** 데이터는 머신 러닝이 학습 모델을 만드는 데 사용하는 것으로, 훈련 데이터가 나쁘다면 실제 현상의 특성을 제대로 반영할 수 없으므로 실제 데이터의 특징이 잘 반영되고 편향되지 않는 훈련 데이터의 확보가 중요함. 또, 학습에 필요한 데이터가 수집되었다면 '훈련 데이터셋'과 '테스트 데이터셋' 용도로 분리해 사용함. 혹은 '훈련 데이터셋'을 또다시 '훈련 데이터셋'과 '검증 데이터셋'으로 분리해서 사용하기도 함. (* 검증 데이터셋을 사용하는 이유는 모델 성능을 평가하기 위함임. 모델 성능의 평가는 테스트 데이터셋에 대한 성능을 가늠할 수 있고, 훈련 데이터셋에 대한 정확도는 높은데 검증 데이터셋에 대한 정확도가 낮다면 훈련 데이터셋에 과적합이 일어났을 가능성이 있으므로 정규화를 하거나 에포크(epoch)를 줄이는 방식으로 과적합을 막을 수 있으므로 모델 성능을 높이는 데 도움을 줄 수 있음.) **모델은 머신 러닝의 학습 단계에서 얻은 최종 결과물로 가설이라고도 함**. (e.g. "입력 데이터의 패턴은 A와 같다." 라는 가정을 머신 러닝에서는 모델이라고 함.) 모델의 학습 절차는


1.   모델(또는 가설) 선택
2.   모델 학습 및 평가
3. 평가를 바탕으로 모델 업데이트

이며 이 세 단계를 반복하며 주어진 문제를 가장 잘 풀 수 있는 모델을 찾음.





### 머신 러닝 학습 알고리즘

*   머신 러닝의 학습 알고리즘으로는 지도 학습, 비지도 학습, 강화 학습이 있음.
*   **지도 학습**은 이름에서 알 수 있듯 **정답이 무엇인지 컴퓨터에 알려주고 학습시키는 방법**임. 지도 학습은 주어진 데이터에 대해 A 혹은 B로 명확한 분류가 가능함.
* 반면 **비지도 학습**은 **정답을 알려주지 않고 특징이 비슷한 데이터를 클러스터링(범주화)하여 예측하는 학습 방법**임. 비지도 학습은 유사도 기반(데이터 간 거리 측정)으로 특징이 유사한 데이터끼리 클러스터링으로 묶어 분류함.
* 강화 학습은 분류할 수 있는 데이터가 있는 것도 아니고, 데이터가 있다고 해도 정답이 없음. 강화 학습은 자신의 행동에 대한 보상을 받으며 학습을 진행함. 보상이 커지는 행동은 자주 하도록 하고, 줄어드는 행동은 덜 하도록 하여 학습을 진행함.

### 딥러닝이란

*   **딥러닝은 인간의 신경망 원리를 모방한 심층 신경망 이론을 기반으로 고안된 머신 러닝 방법의 일종임.** 인간의 뇌가 엄청나게 많은 수의 뉴런(neuron)과 시냅스(synapse)로 구성되어 있는 것에 착안해 컴퓨터에 뉴런과 시냅스 개념을 적용함. **각각의 뉴런은 복잡하게 연결된 수많은 뉴런을 병렬 연산해 기존에 컴퓨터가 수행하지 못했던 음성, 영상 인식 등 처리를 가능케 함.**
IMG_3159.jpg





*   **수상돌기**: 주변이나 다른 뉴런에서 자극을 받아들이고, 이 자극들을 전기적 신호 형태로 세포체와 축삭돌기로 보내는 역할을 함.
*   **시냅스**: 신경 세포들이 이루는 연결 부위로, 한 뉴런의 축삭돌기와 다음 뉴런의 수상돌기가 만나는 부분임.
*   **축삭돌기**: 다른 뉴런(수상돌기)에 신호를 전달하는 기능을 하는 뉴런의 한 부분임. 뉴런에서 뻗어 있는 돌기 중 가장 길며, 한 개만 있음.
*   **축삭말단**: 전달된 전기 신호를 받아 신경 전달 물질을 시냅스 틈새로 방출함.

### 딥러닝 학습 과정
*   데이터 준비부터 모델을 정의하고 사용하는 상위 레벨에서는 딥러닝의 학습 과정도 머신 러닝과 크게 다르지 않음.

1.   **데이터 준비**: 파이토치나 케라스에서 제공하는 이미 전처리된 데이터셋을 사용하거나 캐글 같은 곳의 공개된 데이터를 사용할 수 있음.
2.   **모델 정의**: 모델 정의 단계에서 신경망을 생성함. 일반적으로 은닉층 개수가 많을수록 성능이 좋아지지만 과적합이 발생할 확률이 높음. 즉 은닉층 개수에 따른 성능과 과적합은 서로 상충 관계에 있다 할 수 있음. 따라서 모델 정의 단계에서 신경망을 제대로 생성하는 것이 중요함.
3. **모델 컴파일**: 컴파일 단계에서 활성화 함수, 손실함수, 옵티마이저를 선택함. 이 때 데이터 형태에 따라 다양한 옵션이 가능함.
4. **모델 훈련**: 훈련 단계에서는 한 번에 처리할 데이터양을 지정함. 이 때 한 번에 처리해야 할 데이터양이 많아지면 학습 속도가 느려지고 메모리 부족을 야기할 수 있으므로 적당한 데이터양을 선택하는 것이 중요함. 따라서 전체 훈련 데이터셋에서 일정한 묶음으로 나누어 처리할 수 있는 배치와 훈련의 횟수인 에포크(epoch) 선택이 중요함. (e.g. 훈련 데이터셋 1000개에 대한 배치 크기가 20'이라면 샘플 단위 20개마다 모델 가중치를 한 번씩 업데이트한다는 의미이며 총 50번의 가중치가 업데이트됨. 이 때 에포크가 10이고 배치 크기가 20이라면, 가중치를 50번 업데이트하는 것을 총 10번 반복한다는 의미임. 각 데이터 샘플이 총 10번 사용되는 것이므로 가중치가 총 500번 업데이트됨.)
5. **모델 예측**: 검증 데이터셋을 생성한 모델에 적용해 실제로 예측을 진행해봄. 이 때 예측력이 낮으면 파라미터를 튜닝하거나 신경망 자체를 재설계해야할 수 있음.







*   딥러닝 학습 과정에서 중요한 핵심 구성 요소는 신경망과 역전파임. 딥러닝은 머신 러닝의 한 분야이기는 하지만, 심층 신경망(deep neural network)을 사용한다는 점에서 머신 러닝과 차이가 있음. 은닉층이 두 개 이상인 신경망을 뜻하는 심층 신경망에는 데이터셋의 어떤 특성들이 중요한지 스스로에게 가르쳐줄 수 있는 기능이 있음.

*   가중치 값을 업데이트하기 위한 역전파도 중요함. 특히 역전파 계산 과정에서 사용되는 미분(오차를 각 가중치로 미분)이 성능에 영향을 미치는 주요한 요소라고 할 수 있음. 파이토치 같은 프레임워크를 이요하면 역전파 알고리즘을 자동으로 처리해주므로 딥러닝 알고리즘 구현이 매우 간단해짐. 즉, 파이토치를 이용하면 딥러닝 알고리즘 구현이 간단하고 편리함.

### 딥러닝 학습 알고리즘


*   활용분야에 따라 지도 학습, 비지도 학습, 전이 학습으로 분류되는데 먼저 지도 학습으로 뷴류되는 알고리즘을 살펴보면, 이미지 분류는 이미지 또는 비디오상의 객체를 식별하는 컴퓨터 비전 기술임. 컴퓨터 비전에 가장 많이 사용되는 것이 합**성곱 신경망(Convolutional Neural Network, CNN)**임. 합성곱 신경망은 목적에 따라 이미지 분류, 이미지 인식, 이미지 분할로 분류할 수 있음. 이미지 분류는 이미지를 알고리즘에 입력하면 그 이미지가 어떤 클래스에 속하는지 알려주며, 이미지 인식은 사진을 분석해 그 안에 있는 사물의 종류를 인식함. 그리고 이미지 분할은 영상에서 사물이나 배경 등 객체 간 영역을 픽셀 단위로 구분하는 기술임. (e.g. X-ray, CT(Computer Tomography), MRI(Magnetic Resonance Imaging)를 통한 질병 진단)
*   시계열 데이터를 분류할 때 사용되는 것이 **순환 신경망(Recurrent Neural Network. RNN)**임. 주식 데이터처럼 시간에 따른 데이터가 있을 때 이를 사용하나, 역전파 과정에서 기울기 소멸 문제가 발생하는 단점이 있음. 이러한 문제점을 개선하고자 게이트(gate)를 세 개 추가한 게 바로 **LSTM(Long Short-Term Memory)**임. 망각 게이트(과거의 정보를 잊기 위한 게이트), 입력 게이트(현재 정보를 기억하기 위한 게이트)와 출력 게이트(최종 결과를 위한 게이트)를 도입해 기울기 소멸 문제를 해결했으며 현재 시계열 문제에서 가장 활발히 사용됨.





*   **비지도 학습**으로는 워드 임베팅과 군집이 있음. 자연어, 즉 사람의 언어를 컴퓨터가 이해하고 효율적으로 처리하게 하려면 컴퓨터가 이해할 수 있도록 자연어를 적절히 변환하는 것이 필요한데, 이 때 **워드 임베딩(word embedding)** 기술을 이용해 단어를 벡터로 표현함. (e.g. 번역이나 음성 인식 서비스 등) 워드 임베딩에서는 단어 의미를 벡터화하는 워드두벡터(Word2Vec)와 글로브(GloVe)를 가장 많이 사용함. 군집은 아무런 정보가 없는 상태에서 데이터를 분류하는 방법임. 한 클러스터 안의 데이터는 매우 비슷하게 구성하고 다른 클러스터의 데이터와 구분되도록 나누는 것이 목표임. **군집**은 머신 러닝의 군집과 크게 다르지 않은데, 머신 러닝에서 군집화를 처리할 때 딥러닝과 함께 사용하면 모델 성능을 높일 수 있어 머신러닝 단독으로 군집 알고리즘을 적용하기보다 딥러닝과 함께 사용하면 좋음.
*   **전이 학습(transfer learning)**은 사전에 학습이 완료된 모델을 가지고 원하는 학습에 미세 조정 기법을 이용해 학습시키는 방법임. 사전 학습 모델은 풀고자 하는 문제와 비슷하면서 많은 데이터로 이미 학습이 되어 있는 모델임. VGG, 인셉션, MobileNet과 같은 사전 학습 모델을 사용하면 효율적인 학습이 가능함.

```python
import torch.nn as nn
class MLP(nn.Module):
  def __init__(self):
    super(MLP, self).__init__()
    self.layer1 = nn.Sequential(
        nn.Conv2d(in_channels=3, out_channels=64, kernel_size=5),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2))
    self.layer2 = nn.Sequential(
        nn.Conv2d(in_channels=64, out_channels=30, kernel_size=5),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2))
    self.layer3 = nn.Sequential(
        nn.Linear(in_features=30*5*5, out_features=10, bias=True),
        nn.ReLU(inplace=True))

    def forward(self, x):
      x = self.layer1(x)
      x = self.layer2(x)
      x = x.view(x.shape[0], -1)
      x = self.layer3(x)
      return x
model = MLP()

print("Printing children\n------------------------------")
print(list(model.children()))
print("\n\nPrinting Modules\n------------------------------")
print(list(model.modules()))
```

```python
def MLP(in_features=1, hidden_features=20, out_features=1):
  hidden = nn.Linear(in_features=in_features, out_features=hidden_features, bias=True)
  activation = nn.ReLU()
  output = nn.Linear(input_features=hidden_features, out_features=out_features, bias=True)
  net = nn.Sequential(hidden, activation, output)
  return net
```

```python
import torch
import torchmetrics

preds = torch.randn(10, 5).softmax(dim=-1)
target = torch.randint(5, (10,))

acc = torchmetrics.functional.accuracy(preds, target, task='multiclass', num_classes=5)
```

```python
import torch
import torchmetrics
metric = torchmetrics.Accuracy(task='multiclass', num_classes=5)

n_batches = 10
for i in range(n_batches):
  preds = torch.randn(10, 5).softmax(dim=-1)
  target = torch.randint(5, (10,))
  acc = metric(preds, target)
  print(f"Accuracy on batch {i}: {acc}")

acc = metric.compute()
print(f"Accuracy on all data: {acc}")
```
